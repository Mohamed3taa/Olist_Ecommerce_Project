#### -- ***Handel Missing Values in "products"***

- ***fillna(0) ensures no NaN values remain.***
- ***Then, any 0 values in numeric columns are replaced by their mean.***

```python
# Step 1: Replace all NaN with 0
products = products.fillna(0)

# Step 2: Define Numeric Columns
numeric_cols = [
    'product_name_lenght',
    'product_description_lenght',
    'product_photos_qty',
    'product_weight_g',
    'product_length_cm',
    'product_height_cm',
    'product_width_cm'
]


# Step 3: Replace numeric zeros with mean
for col in numeric_cols:
    mean_value = products[col].mean()
    products[col] = products[col].replace(0, mean_value)

# Step 4: Confirm no missing values remain
products.isnull().sum()
```

- ***Display the mean (average) value for each numeric column***

```python
# Define numeric columns
numeric_cols = [
    'product_name_lenght',
    'product_description_lenght',
    'product_photos_qty',
    'product_weight_g',
    'product_length_cm',
    'product_height_cm',
    'product_width_cm'
]

# Calculate and print the average (mean) for each numeric column
print("---<<< Average (Mean) of Each Numeric Column: >>>---\n")
for col in numeric_cols:
    avg_value = products[col].mean()
    print(f"{col}: {avg_value:.2f}")
```

### ***9. Aggregations***

```python
# Olist Monthly Revenue by Year

# === 1. Define dataset path ===
# Change this to where your Kaggle dataset is saved
olist_path = "/kaggle/input/brazilian-ecommerce"

# === 2. Load datasets ===
orders = pd.read_csv(os.path.join(olist_path, "olist_orders_dataset.csv"))
payments = pd.read_csv(os.path.join(olist_path, "olist_order_payments_dataset.csv"))
customers = pd.read_csv(os.path.join(olist_path, "olist_customers_dataset.csv"))

# === 3. Convert date column to datetime ===
orders['order_purchase_timestamp'] = pd.to_datetime(orders['order_purchase_timestamp'])

# Extract year and month
orders['year'] = orders['order_purchase_timestamp'].dt.year
orders['month'] = orders['order_purchase_timestamp'].dt.month

# === 4. Merge orders with payments ===
order_data = orders.merge(payments, on='order_id', how='left')

# === 5. Revenue by Year/Month ===
revenue_by_year_month = (
    order_data.groupby(['year', 'month'])['payment_value']
    .sum()
    .reset_index()
    .rename(columns={'payment_value': 'total_revenue'})
)

print("Revenue by Year and Month:")
print(revenue_by_year_month.head(), "\n")

# === 6. Average Order Value (AOV) ===
aov = (
    order_data.groupby(['year', 'month'])
    .agg(total_revenue=('payment_value', 'sum'),
         num_orders=('order_id', 'nunique'))
    .reset_index()
)
aov['AOV'] = aov['total_revenue'] / aov['num_orders']

print("Average Order Value (AOV):")
print(aov.head(), "\n")

# === 7. Repeat Customers ===
repeat_customers = (
    orders.groupby('customer_id')['order_id']
    .nunique()
    .reset_index()
    .rename(columns={'order_id': 'num_orders'})
)

repeat_summary = {
    'Total_Customers': repeat_customers.shape[0],
    'Repeat_Customers': (repeat_customers['num_orders'] > 1).sum(),
    'Repeat_Rate (%)': round((repeat_customers['num_orders'] > 1).mean() * 100, 2)
}

print("Repeat Customer Summary:")
print(repeat_summary, "\n")

# === 8. Optional Visualization ===
import matplotlib.pyplot as plt

plt.figure(figsize=(10,5))
for y in revenue_by_year_month['year'].unique():
    subset = revenue_by_year_month[revenue_by_year_month['year'] == y]
    plt.plot(subset['month'], subset['total_revenue'], marker='o', label=f"{y}")

plt.title("Olist Monthly Revenue by Year")
plt.xlabel("Month")
plt.ylabel("Total Revenue (BRL)")
plt.legend()
plt.grid(True)
plt.show()
```

###  Insights: Olist Monthly Revenue by Year

- The revenue trend shows a **steady growth from 2016 to 2017**, indicating expanding operations and customer adoption.  
- **2018 recorded the highest monthly revenues**, consistently exceeding **1 million BRL** in the first eight months — highlighting strong early-year sales performance.  
- A sharp **drop after August 2018** suggests incomplete order/payment data for later months or reduced seasonal sales activity.
